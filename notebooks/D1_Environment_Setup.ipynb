# D1 - Environment Setup

## Objective
The objective of D1 is to set up a reproducible development environment for the Task Management API project, establish the core repository structure, configure dependencies, and verify the FastAPI and ASGI application entry points using Pydantic settings.

## Concepts Covered
- **Python Environment & Runtime Verification**: Verifying the Python version and workspace paths.
- **FastAPI Framework & ASGI Integration**: Initializing the core application instance and middleware.
- **Pydantic Settings & Environment Variables**: Loading application configuration safely (`BaseSettings`).
- **Modular Project Structure**: Organizing code into clean, scalable packages (`app/`, `tests/`, `scripts/`).

## Project Implementation
The application structure is organized into modular packages:
- `app/main.py`: Creates the FastAPI application instance, configures lifespan events, and registers routers.
- `app/config.py`: Defines `Settings` powered by `pydantic_settings` and `PipelineConfig` Pydantic models.
- `app/api/`: Implements API endpoints for tasks, users, and health checks.

In this demonstration, we inspect the Python environment, project layout, configuration, and app initialization using direct imports from the project.

## Demonstration
We perform environment inspection and verify the project modules using Python code cells executed from the repository root.

```python
import sys
from pathlib import Path

# Resolve project root directory regardless of execution context
root_dir = Path.cwd().resolve()
if root_dir.name == "notebooks":
    root_dir = root_dir.parent
if str(root_dir) not in sys.path:
    sys.path.insert(0, str(root_dir))

print(f"Python Version: {sys.version}")
print(f"Executable: {sys.executable}")
print(f"Project Root Directory: {root_dir}")
```
